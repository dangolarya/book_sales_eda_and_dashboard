```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Data visualization:

```python
plt.hist(df['Publishing Year'], bins=10)
plt.title('Distribution of publishing Year')
plt.xlabel('Publishing Years')
plt.ylabel('Frequency')
plt.show()
print('The above histogram shows the distribution of publishing year where the books between the publishing years 2000 and 2025 has highest frequency.')
```

```python
df['genre'].value_counts()
```

```python
counts = df['genre'].value_counts()
plt.bar(counts.index, counts, color='orange')
plt.title('Number of Books of each genre')
plt.xlabel('Genre')
plt.ylabel('No of books')
plt.show()
print('According to the bar diagram, the books of the genre fiction has highest no of books compared to other genre.')
```

```python
unit_sold_per_publisher=df.groupby('Publisher')['units sold'].sum()
```

```python
units_in_m=round(unit_sold_per_publisher/1000000, 2)
units_in_m
```

```python
plt.bar(units_in_m.index, units_in_m)
plt.title('Number of units sold by each publisher')
plt.xticks(rotation=90)
plt.xlabel('Publishers')
plt.ylabel('Units sold in millions')
plt.show()
print('According to the bar diagram, the Amazon Digital Services, Inc. publisher sold the highest number of units compared to other publisher by a very high margin.')
```

```python
author_rating=df['Author_Rating'].value_counts()
author_rating
```

```python
plt.figure(figsize=(9,9))
plt.title('Author rating shown in pie chart')
plt.pie(author_rating, labels=author_rating.index, startangle=90, autopct='%1.1f%%', labeldistance=1.05)
plt.legend()
plt.show()
print('As per the pie chart, 58.7% of the books have been rated as intermediate whereas only 2.8% are rated as novice, 4.8% as famous and 33.7% as excellent.')
```

```python
df
```

```python
grouped_year=df.groupby('Publishing Year')['sale price'].mean().reset_index()
```

```python
plt.figure(figsize=(10,6))
sns.lineplot(x="Publishing Year", y="sale price", data=grouped_year, color="skyblue")
plt.xlabel("Publishing Year")
plt.ylabel("Sale Price")
plt.title("Average Price by Year")
plt.show()
print('According to this line plot, the sale price of the book peaked between the years 1875 and 1912.')
```

```python
plt.xlabel('Sale Price')
plt.ylabel('Units sold')
plt.title('Scatter plot of sale price vs units sold')
plt.scatter(df['sale price'], df['units sold'], c='r', alpha=0.5)
plt.show()
print('According to this scatter plot, we can observe that lesser the sale price of the books, higher the units are sold and vice-versa.')
```

```python
top_authors=df.groupby('Author')['Book_average_rating'].mean().sort_values(ascending=False).head(10).sort_values()
```

```python
plt.xlabel('Average book rating')
plt.ylabel('Authors')
plt.title('Top 10 authors by their average book rating')
plt.barh(top_authors.index, top_authors.values)
plt.show()
print('As per the chart, Bill Watterson is the best author among others according to their average book rating')
```

```python
plt.xlabel('Gross sales')
plt.ylabel('Publisher revenue')
plt.title('Scatter plot of gross sales vs publisher revenue')
plt.scatter(df['gross sales'], df['publisher revenue'], c='r', alpha=0.5)
plt.show()
print('According to this scatter plot, we can see that with the increase gross sales, the publisher revenue also increases. However, in few cases we can observe that increase in gross results in zero publisher revenue.')
```

```python
cols = [
    'units sold',
    'gross sales',
    'publisher revenue',
    'sale price',
    'Book_average_rating'
]

corr = df[cols].corr()

sns.heatmap(corr, annot=True, cmap='Oranges')
plt.title('Correlation Heatmap')
plt.show()
print('The above heatmap shows the correlation between multiple numeric variables where 1 represents positive correlaation, -1 represents negative correlation and 0 represents weak correlation.')
```
